```python
import pandas as pd
```

```python
df = pd.read_parquet("phase3_feature_engineering/feature engineered datasets/playlist.parquet")
```

```python
filtered_df = df[df['num_tracks'] >= 20]
```

```python
import numpy as np
import pandas as pd

# Function to process track_idx_list
def process_track_list(x):
    if isinstance(x, np.ndarray):  
        track_list = x.tolist()  # Convert NumPy array to Python list
    elif isinstance(x, str):  
        import ast
        track_list = ast.literal_eval(x)  # Convert string to list
    elif isinstance(x, list):
        track_list = x  # Already a list, keep it as is
    else:
        return [], []  # Handle unexpected cases

    # Split into masked and last 10 elements
    if len(track_list) > 10:
        return track_list[:-10], track_list[-10:]
    else:
        return [], track_list  # If fewer than 10 elements, store all in last_10

# Apply function row-wise
filtered_df[['tracks_masked', 'tracks_to_predict']] = (
    filtered_df['track_idx_list'].apply(lambda x: pd.Series(process_track_list(x)))
)
```
